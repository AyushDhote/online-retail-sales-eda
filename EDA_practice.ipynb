```python
df_clean = df.copy()
df_clean = df_clean.drop_duplicates()
```

```python
print("Original rows:",len(df))
print("Rows after removing duplicates:",len(df_clean))
print("duplicates remaining:",df_clean.duplicated().sum())
```

```python
print("Negative quantity:",(df_clean["Quantity"]<0).sum())
print("zero qunatity:",(df_clean["Quantity"]==0).sum())
print("\n")
print("Negative unitprice:",(df_clean["UnitPrice"]<0).sum())
print("zero qunatity:",(df_clean["UnitPrice"]==0).sum())
```

```python
print("neagative quantity:")
print(df_clean[df_clean["UnitPrice"]<0])
print("\nZero quantity:")
print(df_clean[df_clean["UnitPrice"]==0].head(5))
print("\nNegative quantity:")
print(df_clean[df_clean["Quantity"]<0].head(5))
```

```python
df_clean["InvoiceDate"]= pd.to_datetime(df_clean["InvoiceDate"])
print(df_clean["InvoiceDate"].dtype)
print("\nDate range: ")
print(df_clean["InvoiceDate"].min())
print(df_clean["InvoiceDate"].max())
```

```python
df_clean["Revenue"]= df_clean["UnitPrice"]*df_clean["Quantity"]
print("total revenue:",df_clean["Revenue"].sum())
print("average of revenue:",df_clean["Revenue"].mean())
print("maximum revenue:",df_clean["Revenue"].max())
print("minimum revenue",df_clean["Revenue"].min())
```

```python
df_clean["Month"] = df_clean["InvoiceDate"].dt.to_period("M")
monthly_revenue = df_clean.groupby("Month")["Revenue"].sum()
print(monthly_revenue)
```

```python
import matplotlib.pyplot as plt
monthly_revenue.plot(kind="bar",figsize=(12,5))
plt.title("Monthly revenue trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.yticks(rotation=-45)
plt.tight_layout()
plt.show()
```

```python
top_product = (df_clean.groupby("Description")["Revenue"].sum().sort_values(ascending=False).head(5)
              )
print(top_product)
```
